```python
bene_columns = [
    'BENE_ID',
    'STATE_CODE',
    'COUNTY_CD',
    'ZIP_CD',
    'BENE_BIRTH_DT',
    'SEX_IDENT_CD',
    'BENE_RACE_CD',
    'BENE_DEATH_DT',
    'AGE_AT_END_REF_YR'
]

beneficiary_slim = beneficiary[bene_columns].copy()

# Create death flag
beneficiary_slim['DEATH_FLAG'] = beneficiary_slim['BENE_DEATH_DT'].apply(lambda x: 0 if pd.isna(x) else 1)

print(beneficiary_slim.shape)
beneficiary_slim.head()
```

```python
outpatient_columns = [
    'BENE_ID',
    'CLM_ID',
    'CLM_FROM_DT',
    'CLM_THRU_DT',
    'CLM_PMT_AMT',
    'CLM_TOT_CHRG_AMT',
    'NCH_PRMRY_PYR_CLM_PD_AMT',
    'PRVDR_NUM',
    'PRVDR_STATE_CD',
    'PRNCPAL_DGNS_CD',
    'PTNT_DSCHRG_STUS_CD'
]

outpatient_slim = outpatient[outpatient_columns].copy()
print(outpatient_slim.shape)
outpatient_slim.head()
```

```python
# Inpatient duplicates
print("Inpatient duplicates:", inpatient_slim.duplicated(subset=['BENE_ID', 'CLM_ID']).sum())

# Outpatient duplicates
print("Outpatient duplicates:", outpatient_slim.duplicated(subset=['CLM_ID', 'CLM_FROM_DT', 'CLM_TOT_CHRG_AMT']).sum())

# Beneficiary duplicates
print("Beneficiary duplicates:", beneficiary_slim.duplicated(subset=['BENE_ID']).sum())
```

```python
# Remove duplicates - keep first occurrence
inpatient_clean = inpatient_slim.drop_duplicates(subset=['BENE_ID', 'CLM_ID'], keep='first')

outpatient_clean = outpatient_slim.drop_duplicates(subset=['CLM_ID', 'CLM_FROM_DT', 'CLM_TOT_CHRG_AMT'], keep='first')

# Check how many rows we have now
print("Inpatient before:", inpatient_slim.shape[0], "→ after:", inpatient_clean.shape[0])
print("Outpatient before:", outpatient_slim.shape[0], "→ after:", outpatient_clean.shape[0])
```

```python
inpatient_clean = inpatient_slim.drop_duplicates(subset=['BENE_ID', 'CLM_ID'], keep='first').copy()

outpatient_clean = outpatient_slim.drop_duplicates(subset=['CLM_ID', 'CLM_FROM_DT', 'CLM_TOT_CHRG_AMT'], keep='first').copy()
```

```python
# See current data types of all columns
print("=== INPATIENT ===")
print(inpatient_clean.dtypes)

print("\n=== OUTPATIENT ===")
print(outpatient_clean.dtypes)

print("\n=== BENEFICIARY ===")
print(beneficiary_slim.dtypes)
```

```python
# Fix dates - Inpatient
inpatient_clean['CLM_FROM_DT'] = pd.to_datetime(inpatient_clean['CLM_FROM_DT'])
inpatient_clean['CLM_ADMSN_DT'] = pd.to_datetime(inpatient_clean['CLM_ADMSN_DT'])
inpatient_clean['NCH_BENE_DSCHRG_DT'] = pd.to_datetime(inpatient_clean['NCH_BENE_DSCHRG_DT'])

# Fix dates - Outpatient
outpatient_clean['CLM_FROM_DT'] = pd.to_datetime(outpatient_clean['CLM_FROM_DT'])
outpatient_clean['CLM_THRU_DT'] = pd.to_datetime(outpatient_clean['CLM_THRU_DT'])

beneficiary_slim['BENE_BIRTH_DT'] = pd.to_datetime(beneficiary_slim['BENE_BIRTH_DT'])
beneficiary_slim['BENE_DEATH_DT'] = pd.to_datetime(beneficiary_slim['BENE_DEATH_DT'])

print(beneficiary_slim.dtypes)
print(inpatient_clean.dtypes)
print(outpatient_clean.dtypes)
```

```python
zero_days = inpatient_clean[inpatient_clean['CLM_UTLZTN_DAY_CNT'] == 0].copy()
zero_days['date_diff'] = (zero_days['NCH_BENE_DSCHRG_DT'] - zero_days['CLM_ADMSN_DT']).dt.days
print(zero_days['date_diff'].value_counts())
```

```python
# Only remove where BOTH utilization days = 0 AND date difference = 0
inpatient_clean['date_diff'] = (inpatient_clean['NCH_BENE_DSCHRG_DT'] - inpatient_clean['CLM_ADMSN_DT']).dt.days

print("Before:", inpatient_clean.shape[0])
inpatient_clean = inpatient_clean[~((inpatient_clean['CLM_UTLZTN_DAY_CNT'] == 0) & (inpatient_clean['date_diff'] == 0))]
print("After:", inpatient_clean.shape[0])
```

```python
# Update CLM_UTLZTN_DAY_CNT with date_diff where it was wrong
inpatient_clean['CLM_UTLZTN_DAY_CNT'] = inpatient_clean['date_diff']

# Now drop date_diff since we don't need it anymore
inpatient_clean = inpatient_clean.drop(columns=['date_diff'])

print(inpatient_clean['CLM_UTLZTN_DAY_CNT'].value_counts().head(10))
```

```python
inpatient_clean['PAYMENT_GAP'] = inpatient_clean['CLM_TOT_CHRG_AMT'] - inpatient_clean['CLM_PMT_AMT']

print(inpatient_clean['PAYMENT_GAP'].describe())
```

```python
inpatient_clean = inpatient_clean.drop(columns=['PAYMENT_GAP'])
print(inpatient_clean.columns.tolist())
```

```python
outpatient_clean['PAYMENT_GAP'] = outpatient_clean['CLM_TOT_CHRG_AMT'] - outpatient_clean['CLM_PMT_AMT']
print(outpatient_clean['PAYMENT_GAP'].describe())
```

```python
outpatient_clean = outpatient_clean.drop(columns=['PAYMENT_GAP'])
print(outpatient_clean.columns.tolist())
```
